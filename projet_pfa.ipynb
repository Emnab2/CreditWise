```python
from sklearn.preprocessing import OrdinalEncoder

statuts_test = [['Désinscrit'], ['Enregistré'], ['Inscrit']]

encoder = OrdinalEncoder(categories=[['Désinscrit', 'Enregistré', 'Inscrit']])
encoded_data = encoder.fit_transform(statuts_test)

print(f"Encoded Ordinal Data: {encoded_data}")
```
