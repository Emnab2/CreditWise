```python
import pandas as pd
data=pd.read_csv("https://raw.githubusercontent.com/jakevdp/data-CDCbirths/master/births.csv")
data
```

```python

data["day"].unique()
```

```python
data=data[(data['day'] <= 31)&(data['day'].isnull()==False)]
```

```python
data["day"].unique()
```

```python
data['date']=pd.to_datetime(data[["year","month","day"]],errors="coerce")
```

```python
data
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

```python
data['day_name']=data['date'].dt.day_name()
data
```

```python
a=1988
def decade(x):
    y=str(x)[2]+"0"
    return y
decade(a)
```

```python
data['decade']=data["year"].apply(decade)
data
```

```python
data_final=pd.DataFrame(data.groupby(["decade","day_name"])["births"].mean())
data_final=data_final.unstack(level=0)
data_final
```

```python
data_final=data_final.reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
data_final
```

```python
data_final.plot(kind='bar', subplots=False)
```
