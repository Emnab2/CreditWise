```python
import os
import pandas as pd
files =[file for file in os.listdir(r'C:\Users\MSI\Downloads\PFA emna\project')]
for file in files :
    print(file)
```

```python
all_data=pd.DataFrame()# créer un objet base de données en en doit le remplire
for file in files :
    current_data=pd.read_csv( r'C:\Users\MSI\Downloads\PFA emna\project'+'/'+ file)
    all_data = pd.concat([all_data,current_data])

all_data.head()
all_data.shape
#all_data.describe()
all_data
```

```python
all_data.dropna(inplace=True)
```

```python
all_data.isnull().sum()
```

```python
all_data
```

```python
print(all_data['Order ID'].unique())
```

```python
print(all_data['Quantity Ordered'].unique())
```

```python
print(all_data['Price Each'].unique())
```

```python
all_data = all_data[(all_data['Quantity Ordered'] != 'Quantity Ordered') & (all_data['Price Each'] != 'Price Each')]
all_data
```

```python
print(all_data['Price Each'].unique())
```

```python
print(all_data['Quantity Ordered'].unique())
```

```python
all_data['Quantity Ordered'] = all_data['Quantity Ordered'].astype(float)
all_data['Price Each'] = all_data['Price Each'].astype(float)
```

```python
all_data['montant_tot'] = all_data['Quantity Ordered'] * all_data['Price Each']
all_data
```

```python
all_data['Order Date'] = pd.to_datetime(all_data['Order Date'])
all_data['Month'] = all_data['Order Date'].dt.month
all_data
```

```python
revenue_mois = pd.DataFrame(all_data.groupby('Month')['montant_tot'].sum())
print(revenue_mois)
```

```python
print(revenue_mois.idxmax())
```

```python
revenue_mois.plot(kind='bar', subplots=False)
```

```python
def cité(adresse):
    return adresse.split(',')[1]
all_data['City'] = all_data['Purchase Address'].apply(cité)
all_data
```

```python
print(all_data['City'].unique())
```

```python
commande_par_ville = pd.DataFrame(all_data.groupby('City')['City'].count())
commande_par_ville.columns = ['nombre_commandes']
print(commande_par_ville)
```

```python
ville_max = commande_par_ville['nombre_commandes'].idxmax()
print(ville_max)
```

```python
commande_par_ville.plot(kind='bar', subplots=False)
```
